```python
# Reading/Loading the image :-

import cv2

image = cv2.imread("cat pic.jpg")

if image is None:
    print("Error: Image is not found")

else:
    print("Image loaded successfully")
```

```python
# Displaying the image :-

image = cv2.imread("cat pic.jpg")

if image is not None:
    cv2.imshow("Image showing", image)   # open the window with title "Image showing"
    print("image displayed")
    cv2.waitKey(0)  # wait for a key
    cv2.destroyAllWindows()   # close the window

else:
     print("could not load the image")
```

```python
# Saving the image :-

image = cv2.imread("cat pic.jpg")


if image is not None:
    success = cv2.imwrite("cat_output_pic.jpg", image)

    if success:
        print("Image saved successfully as 'cat_output_pic.jpg'")

    else:
        print("Failed to save an image")

else:
    print("Error: Could not load image")
```

```python
# checking for the shape of an image :-

if image is not None:
    h,w,c = image.shape

    print(f"Image loaded :\nHeight: {h}\nWidth: {w}\nChannels: {c}")

else:
    print("Could not load image")
```

<h2> Grey-scale conversion </h2>

```python
# we use "cvtColor" function -> to change the 'color_type' of an image (e.g., colorful -> grey-scale)

image = cv2.imread("cat pic.jpg")

if image is not None:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    cv2.imshow("Grayscale Image", gray)
    print("Grayscale image generated")
    cv2.waitKey(0)
    cv2.destroyAllWindows()

else:
    print("Could not upload the image")
```
